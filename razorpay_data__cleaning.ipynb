## Step 3: Fixing Inconsistent Formatting
Standardize categorical columns : payment_method, kyc_status, currency, and city using contains-based cleaning functions.

```python
print("Before:", razorpay['payment_method'].value_counts())
print("Unique count:", razorpay['payment_method'].nunique())
```

```python
razorpay['payment_method'].value_counts()

def clean_payment_method(value):
  if value == '' or value == None:
    return 'Unknown'
  v = str(value).strip().lower().replace(' ', '')

  if 'upi' in v or 'upl' in v : return 'UPI'
  if 'credit' in v or 'credt' in v: return 'Credit Card'
  if 'debit' in v: return 'Debit Card'
  if 'net' in v or 'banking' in v: return 'Net Banking'
  if 'wallet' in v: return 'Wallet'
  if 'bnpl' in v or 'buynow' in v: return 'BNPL'
  if 'emi' in v: return 'EMI'
  return value

  #apply the function to every row
razorpay['payment_method'] = razorpay['payment_method'].apply(clean_payment_method)

#check result
print(razorpay['payment_method'].value_counts())
```

```python
razorpay['kyc_status'].value_counts()
def clean_kyc_status(value):

  if value == '' or value == None:
    return 'Unknown'

  v = str(value).strip().lower().replace('_','').replace(' ', '')

  if 'notstarted' in v: return 'Not Started'
  if 'verifed' in v or 'verified' in v or 'verifeid' in v: return 'Verified'
  if 'submit' in v: return 'Submitted'
  if 'startd' in v or 'started' in v: return 'Started'
  return value

razorpay['kyc_status'] = razorpay['kyc_status'].apply(clean_kyc_status)

print(razorpay['kyc_status'].value_counts())
```

```python
razorpay['currency'].value_counts()
def clean_currency(value):
  if value == '' or value == None:
    return 'Unknown'
  v = str(value).strip().lower().replace(' ','')
  if v in ['inr', 'rs', 'indianrupee', '₹']: return 'INR'

  return value

razorpay['currency'] = razorpay['currency'].apply(clean_currency)
razorpay['currency'].value_counts()
```

```python
print(sorted(razorpay['city'].unique()))
print(razorpay['city'].nunique())
```

```python
print("Before:", razorpay['city'].nunique(), "unique values")

def clean_city(value):
  if value == '' or value is None: return 'Unknown'
  v = str(value).strip().lower()
  if v in ['mumbai', 'mombai']: return 'Mumbai'
  if v in ['delhi', 'dehli']: return 'Delhi'
  if v in ['bengaluru', 'bangalore', 'bangaluru']: return 'Bengaluru'
  if v in ['chennai', 'channai']: return 'Chennai'
  if v in ['kolkata', 'calcutta']: return 'Kolkata'
  if v in ['hyderabad', 'hydrabad']: return 'Hyderabad'
  if v in ['pune', 'poona']: return 'Pune'
  if v in ['ahmedabad', 'ahemdabad']: return 'Ahmedabad'
  if v == 'jaipur': return 'Jaipur'
  if v == 'lucknow': return 'Lucknow'
  if v == 'indore': return 'Indore'
  if v == 'kochi': return 'Kochi'
  if v == 'bhopal': return 'Bhopal'
  if v == 'chandigarh': return 'Chandigarh'
  if v == 'surat': return 'Surat'
  return 'Unknown'

razorpay['city'] = razorpay['city'].apply(clean_city)
print("After:", razorpay['city'].nunique(), "unique values")
print(razorpay['city'].value_counts())
```

```python
#Correcting Data Types
# monthly_gmv_inr has text mixed in — convert to number
# errors='coerce' means: if value cannot be converted, make it NaN
razorpay['monthly_gmv_inr'] = pd.to_numeric(razorpay['monthly_gmv_inr'],errors='coerce')

print(razorpay['monthly_gmv_inr'].dtype)
```

```python
#fixing timestamp columns
ts_cols = ['ts_landing_page_view', 'ts_signup_complete',
           'ts_kyc_started', 'ts_kyc_submitted',
           'ts_kyc_verified', 'ts_api_key_generated',
           'ts_first_test_txn', 'ts_first_live_txn',
           'ts_active_merchant', 'ts_premium_upgrade']
for col in ts_cols:
  razorpay[col] = pd.to_datetime(razorpay[col], errors='coerce')

for col in ts_cols:
  print(razorpay[col].dtypes)
```

```python
ts_cols = ['ts_landing_page_view', 'ts_signup_complete',
           'ts_kyc_started', 'ts_kyc_submitted',
           'ts_kyc_verified', 'ts_api_key_generated',
           'ts_first_test_txn', 'ts_first_live_txn',
           'ts_active_merchant', 'ts_premium_upgrade']

current_time = pd.Timestamp.now()
for col in ts_cols:
    print(col, (razorpay[col] > current_time).sum())

for col in ts_cols:
  razorpay.loc[razorpay[col] > current_time, col] = pd.NaT

for col in ts_cols:
    print(col, (razorpay[col] > current_time).sum())
```

## Step 6 : Handling Outliers
Identify negative GMV values and extreme outliers using IQR method.


```python
#identifying outliers

print(f"negitive gmv_inr {(razorpay['monthly_gmv_inr'] < 0). sum()}")

# Problem 2 — look at the top 10 highest values
print()
print("Top 10 GMV values:")
print(razorpay['monthly_gmv_inr'].nlargest(10))

# Also check using IQR method — standard way to detect outliers
Q1 = razorpay['monthly_gmv_inr'].quantile(0.25)
Q3 = razorpay['monthly_gmv_inr'].quantile(0.75)
IQR = Q3 - Q1

upper_limit = Q3 + 1.5 * IQR
lower_limit = Q1 - 1.5 * IQR

print()
print(f"Lower limit: ₹{lower_limit:,.0f}")
print(f"Upper limit: ₹{upper_limit:,.0f}")
print(f"Outliers above upper limit: {(razorpay['monthly_gmv_inr'] > upper_limit).sum()}")
print(f"Outliers below lower limit: {(razorpay['monthly_gmv_inr'] < lower_limit).sum()}")
```

```python
# Are negatives and high outliers the same rows or different?
negative_rows = razorpay['monthly_gmv_inr'] < 0
high_outliers = razorpay['monthly_gmv_inr'] > upper_limit

print("Negative GMV rows:", negative_rows.sum())
print("High outlier rows:", high_outliers.sum())
print("Overlap between both:", (negative_rows & high_outliers).sum())
```

```python
#removing negitve values
razorpay.loc[razorpay['monthly_gmv_inr'] < 0,'monthly_gmv_inr'] = np.nan
print(f"negitive gmv_inr {(razorpay['monthly_gmv_inr'] < 0). sum()}")
```

```python
#removing outliers using IQR
razorpay.loc[razorpay['monthly_gmv_inr'] > upper_limit,'monthly_gmv_inr'] = upper_limit
print(f"negitive gmv_inr {(razorpay['monthly_gmv_inr'] > upper_limit). sum()}")
```
